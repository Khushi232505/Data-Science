## List 

A list in Python is an ordered, mutable collection of elements. It can contain elements of different types.

```python
# Empty list
my_list = []

# Mixed data types
mixed_list = [1, "Hello", 3.14, True]
```

```python
l = [3, 5, 23, 3]
```

```python
l[0] 
```

```python
l[1] = 2
```

```python
print(l)
```

```python
l.append(9) #Adds an element x to the end of the list.
```

```python
l
```

```python
l2 = [5,7]
l.extend(l2) # Extends the list by appending all elements from an iterable.
```

```python
l
```

```python
l.insert(3,"khushi") #Inserts x at the specified index.
```

```python
l
```

```python
l.remove(3) #Removes the first occurrence of x in the list.
```

```python
l
```

```python
l.pop() #Removes and returns the element at index (last element if index is not provided).
```

```python
l
```

```python
l.index(23) #Returns the index of the first occurrence of x.
```

```python
l.insert(2,5)
```

```python
l
```

```python
l.count(5) #Returns the number of times x appears in the list.
```

```python
l.sort() #Sorts the list in ascending order.
```

```python
l
```

```python
l.reverse() #Reverses the order of the list
l
```

```python
l3 = l.copy() #Returns a shallow copy of the list.
l3
```

```python
l3.clear() # Removes all elements from the list
l3
```
