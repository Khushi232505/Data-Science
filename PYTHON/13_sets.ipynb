## Sets & Set Methods
A set in Python is an unordered, mutable, and unique collection of elements. It does not allow duplicate values.


## Key Properties of Sets:
Unordered: No indexing or slicing.

Unique Elements: Duplicates are automatically removed.

Mutable: You can add or remove elements.

```python
s = set()
type(s)
```

```python
# Creating a set from a list
unique_numbers = set([1, 2, 2, 3, 4, 4, 5])
print(unique_numbers)  # {1, 2, 3, 4, 5}
```

```python
s= {1, "Hello", 3.14, True, 4, 6, 1}
print(s)
type(s)
```

```python
s.add(2) # Adds an element x to the set.
s
```

```python
s.remove(3.14) #Removes x from the set (raises an error if not found).
s
```

```python
s.discard(5) #Removes x from the set (does not raise an error if not found).
s
```

```python
s.pop() #Removes and return the random value
```

```python
s
```

```python
s1 = s.copy() #Returns a shallow copy of the set.
s1
```

```python
s.update([2,5,9]) #Adds multiple elements from an iterable.
```

```python
s
```

```python
s.clear()
s
```

## Set Operations


```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
```

```python
set1.union(set2) #print(set1 | set2)  
```

```python
set1.intersection(set2) #print(set1 & set2) 
```

```python
set1.difference(set2) #in set1 but not in set2
# set1-set2
```

```python
set2-set1
```

```python
set1.symmetric_difference(set2) #Returns a set with elements in either set1 or set2, but not both.
```

```python
print(set1 ^ set2)                     
```

```python
s = {3,4}
s.issubset(set1)
```

```python
set1.issuperset(s)
```
