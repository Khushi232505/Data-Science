## Functions

A **function** is a reusable block of code that performs a specific task.


## Defining a function

```python
def greet():
    return ("hello")
```

## Calling a function

```python
greet()
```

## Function with parameters

```python
def greet(name):
    return ("hello", name)
```

```python
print(greet("khushi"))
print(greet("gaurangi"))
```

## Multiple parameters

```python
def add(a,b):
    print(a+b)
```

```python
add(2,4)
```

```python
def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old")
```

introduce("khushi", 21)

```python
def add(a, b):
    print(a + b)

x = add(10, 20)

print(x)
```

```python
def add(a, b):
    return a + b

x = add(10, 20)

print(x)
```

## Default parameters


```python
def add(a,b = 5):
    return a+b
```

```python
print(add(2))
```

```python
def add(a,b):
    return a+b

def subtract(a,b):
    return a-b

def multiplication(a,b):
    return a*b

def division(a,b):
    return a/b

def modulus(a,b):
    return a%b
```

```python
print(add(2,3))
print(subtract(9,5))
print(multiplication(4,2))
print(division(10,2))
print(modulus(11,2))
```

```python
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

is_even(2)
```
