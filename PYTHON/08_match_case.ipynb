## Match-Case in python

In Python, match / case is structural pattern matching, introduced in Python 3.10. It’s similar to switch in other languages, but more powerful.

```python
code = 200
```

```python
match code:
    case 200:
        print("Ok")
        print("good job")
    case 400:
        print("bad response")
    case 404: 
        print("not found")
print("end of program")
```

```python
day = 3
```

```python
match day:
    case 1:
        print("monday")
    case 2:
        print("tuesday")
    case 3: 
        print("wednesday")
    case _: #Here, _ works like the default case.
        print("invalid day")
```
