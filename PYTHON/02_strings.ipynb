# What is a String?

A string is a sequence of characters enclosed in single (') or double (") quotes.

```python
greetings = "hello"
```

# Multiline Strings

Use triple quotes (''' or """) for multiline text.

```python
message = """This is
a multiline
string."""

print(message)
```
