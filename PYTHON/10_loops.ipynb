## Loops in Python

## 1. For Loop

Used to iterate over sequences like lists, tuples, and strings.

```python
l = [3,5,6,67]
```

```python
for item in l:
    print(item)
```

```python
for i in range(1,11):
    print(i)
```

## 2. While Loop

Runs as long as a condition is True.

```python
count = 0
while count < 3:
    print(count)
    count += 1
```

```python
i = 0
while i <12:
    print(i)
    i = i+1
```

## 3. Loop Control Statements
break → Exits the loop.

continue → Skips to the next iteration.

pass → Does nothing (used as a placeholder).

```python
for i in range(0,10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(0,10):
    if i == 5:
        continue
    print(i)
```

```python
for i in range(0,10):
    if i == 5:
        pass
    print(i)
```
